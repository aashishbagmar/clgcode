```python
## DL LAB ASSIGNMENT 5-Design and implement Deep Convolutional GAN to generate images of faces/digits from a set of given images.

import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("sales_data_sample.csv", encoding='latin1')

# Display first rows
print(df.head())

# Basic info
print(df.info())

# Check missing values
print(df.isnull().sum())

# Drop missing values
df = df.dropna()

# Convert ORDERDATE to datetime
df['ORDERDATE'] = pd.to_datetime(df['ORDERDATE'])

# Create YEAR column
df['YEAR'] = df['ORDERDATE'].dt.year

# Group by PRODUCTLINE
product_sales = df.groupby('PRODUCTLINE')['SALES'].sum()

# Bar Chart
product_sales.plot(kind='bar')
plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Sales")
plt.show()

# Group by YEAR
year_sales = df.groupby('YEAR')['SALES'].sum()

# Line Chart
year_sales.plot(kind='line', marker='o')
plt.title("Sales Trend by Year")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.show()

# Group by STATUS
status_sales = df.groupby('STATUS')['SALES'].sum()

# Pie Chart
status_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales by Status")
plt.ylabel('')
plt.show()
```
